```python
import pandas as pd
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

# 1. CSV dataset
df =pd.read_csv("data_analysis_project/Hr_Employe_Attrition.csv")


# 2. MySQL connection
connection_url = URL.create(
    drivername="mysql+pymysql",
    username="root",
    password="shivam@1734",
    host="localhost",
    port=3306,
    database="hr_analysis"
)

engine = create_engine(connection_url)

print("MySQL connected successfully")

# 3. Upload Python DataFrame → MySQL table
df.to_sql(
    name="Hrtable",
    con=engine,
    if_exists="replace",
    index=False
)

print("Dataset uploaded successfully")

# 4. Read table back from MySQL
df_sql = pd.read_sql("SELECT * FROM Hrtable", engine)

df_sql.head()
```
